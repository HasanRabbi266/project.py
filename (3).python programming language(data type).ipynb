```python
#range
x= range(10) #1st case of this function means
```

```python
x #it's value is (0-9) 
```

```python
for i in x:
    print(i)
```

```python
# uses of 2nd case
y =range(5,20) #(5 to 19)
for i in y:
    print(i)
```

```python
#uses of 3rd case
z=range (5,20,3) #increase 3
for i in z:
    print(i)
```

```python
#uses of 4th case
h=range(20,5,-3)#reverse
for i in h:
    print(i)
```
