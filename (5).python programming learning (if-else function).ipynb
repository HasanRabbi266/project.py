conditional statement: if,else,elif

if else statements in python

```python
x = 50
y = 100
if x > y:
    print('x is greater than y')
else :
    print('out!')
```

```python
x = 50
y = 100
if x > y:
    print('x is greater than y')
if y > x:
    print('y is greater than x')
else :
    print('out!')
```

```python
x = 50
y = 100
if x > y:
    print('x is greater than y')
elif y > x:
    print('y is greater than x')
    if x != y:
        print('not equal')
else :
    print('out!')
```
